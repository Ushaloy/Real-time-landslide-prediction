```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
df8=pd.read_csv("/content/drive/MyDrive/energy data/energy data new/soil moisture/Soil_moisture_1sec_01.CSV", skiprows=10)
```

```python
# Assuming columns: Time, CH1
time = df8.iloc[:,0]   # first column = time
voltage = df8.iloc[:,1]  # second column = channel data

# Plot
plt.figure(figsize=(10,5))
plt.plot(time, voltage, label="energy")
plt.xlabel("Time (s)")
plt.ylabel("Current (A)")
plt.title("Soil Moisture sensor energy consumption")
plt.legend()
plt.grid(True)
plt.show()
```

```python
df2=pd.read_csv("/content/drive/MyDrive/energy data/energy data new/Node/Node_4sec_03.CSV", skiprows=10)
```

```python
print(df2.head())

# Assuming columns: Time, CH1
time = df2.iloc[:,0]   # first column = time
voltage = df2.iloc[:,1]  # second column = channel data

# Plot
plt.figure(figsize=(10,5))
plt.plot(time, voltage, label="Energy")
plt.axhline(voltage.max(), color="red", linestyle="--", label="Max")
plt.axhline(voltage.min(), color="purple", linestyle="--", label="Min")
plt.xlabel("Time (s)")
plt.ylabel("Current (A)")
plt.title("Node energy consumption")
plt.legend()
plt.grid(True)
plt.show()
```

```python
v_pp = voltage.max() - voltage.min()
v_rms = np.sqrt(np.mean(voltage**2))

print(f"Peak-to-Peak: {v_pp:.3f} V")
print(f"RMS: {v_rms:.3f} V")
```
