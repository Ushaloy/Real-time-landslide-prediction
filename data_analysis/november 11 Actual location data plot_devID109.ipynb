Filter the `df` dataset to include `devID`s 106, 108, and 109, then for each of these `devID`s, select the first 1200 rows to create a new dataset named `df_3`, retaining all features, and save `df_3` as a CSV file named `df_3.csv`.

## Filter and Sample Data

### Subtask:
Filter the `df` dataset to include only `devID`s 106, 108, and 109. Then, for each of these selected `devID`s, take the first 1200 rows to create the `df_3` dataset, retaining all features.


**Reasoning**:
First, I will filter the original DataFrame `df` to include only `devID`s 106, 108, and 109. Then, for each of these selected `devID`s, I will take the first 1200 rows and concatenate them into a new DataFrame called `df_3`.



```python
selected_devIDs = [106, 108, 109]
df_filtered_devIDs = df[df['col_1'].isin(selected_devIDs)] # Assuming 'col_1' is devID after reloading

sampled_dfs = []
for dev_id in selected_devIDs:
    df_subset = df_filtered_devIDs[df_filtered_devIDs['col_1'] == dev_id] # Assuming 'col_1' is devID
    sampled_dfs.append(df_subset.head(1200))

df_3 = pd.concat(sampled_dfs, ignore_index=True)
print("Shape of df_3:", df_3.shape)
display(df_3.head())
```

**Reasoning**:
The subtask requires saving the `df_3` DataFrame as a CSV file named `df_3.csv`. This is the final step to complete the current subtask.



```python
df_3.to_csv('df_3.csv', index=False)
print("df_3.csv saved successfully.")
```
