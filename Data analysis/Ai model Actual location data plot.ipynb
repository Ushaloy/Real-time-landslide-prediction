```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_ai = df[(df['devID'] == 109) & (df['geo'] >= 0) & (df['geo'] <= 35)].copy()
display(df_ai.head())
df_ai['timestamp'] = pd.to_datetime(df_ai['timestamp'], errors='coerce')
#df1 = df1.sort_values(by='timestamp')
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_ai, x='timestamp', y='iso_risk')
plt.title('Isolatio forest during critical')
plt.xlabel('Timestamp')
plt.ylabel('Alert level')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
