```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#df=pd.read_csv("/content/drive/MyDrive/Landslide project/Data from Actual Location.csv")
df = pd.read_csv("/content/Landslide.csv")
```

```python
df
```

```python
df1 = df[['timestamp','devID','soil', 'rain', 'temp', 'humi', 'geo']]
display(df1.head())
```

```python
df1
```

```python
df1['timestamp'] = pd.to_datetime(df1['timestamp'], errors='coerce')
df1 = df1.sort_values(by='timestamp')

# Filter
df1_filtered = df1[df1['devID'] != 105]
df1_filtered = df1_filtered[df1_filtered['devID'] != 104]

plt.figure(figsize=(12, 6))
sns.lineplot(data=df1_filtered, x='timestamp', y='soil', hue='devID',palette=['red', 'green', 'blue', 'black', 'orange'])
plt.title('Soil Moisture Over Time ')
plt.xlabel('Timestamp')
plt.ylabel('Soil Moisture')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
